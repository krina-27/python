# numpy array and basics

```python
import numpy as np
```

## Creating array from list

```python
arr_1d = np.array([1, 2, 3, 4, 5])
print("end=1d array:", arr_1d)


arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("end=2d array:", arr_2d)
```

## List vs numpy array

```python
py_list= [1,2,3,4,5]
print("multiplication of list with 2:", py_list*2)

py_array = np.array(py_list)
print("multiplication of array with 2:", py_array*2)
```

```python
import time

start = time.time()
py_list = [i * 2 for i in range(1000000)]
print("\n List operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy operation time: ", time.time() - start)
```

## creating array from scratch

```python
zeros = np.zeros((3, 4))
print("\n 3x4 array of zeros:\n", zeros)

ones = np.ones((2, 3))
print("\n 2x3 array of ones:\n", ones)

full = np.full((2, 2), 7)
print("\n 2x2 array filled with 7:\n", full)

random_array = np.random.random((3, 3))
print("\n 3x3 array of random numbers:\n", random_array)

sequence = np.arange(0, 10, 2)
print("\n Sequence from 0 to 10 with step 2:\n", sequence)
```

```python
vector = np.array([1, 2, 3])
print("Vector: ", vector)

matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])
print("Matrix: ", matrix)

tensor = np.array([[[1, 2], [3, 4]], 
                   [[5, 6], [7, 8]]])
print("Tensor: ", tensor)
```

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])
print("Shape ", arr.shape)
print("Dimension ", arr.ndim)
print("Size ", arr.size)
print("DType ", arr.dtype)
```

## Array Reshaping

```python
arr = np.arange(12)
print("Original array ", arr)

reshaped = arr.reshape((3, 4))
print("\n Reshaped array ", reshaped)

flattened = reshaped.flatten()
print("\n Flattened array ", flattened)

# ravel (returns view, instead of copy)
raveled = reshaped.ravel()
print("\n raveled array ", raveled)

# Transpose
transpose = reshaped.T
print("\n Transposed array ", transpose)
```
